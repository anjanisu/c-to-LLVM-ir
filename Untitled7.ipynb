````python
import os
import subprocess
import pandas as pd

# =========================
# CONFIGURATION
# =========================
DATASET_PATH = "C"      # TheAlgorithms/C repo
IR_PATH = "ir_output"
os.makedirs(IR_PATH, exist_ok=True)

MAX_FILES = 100   # 🔥 You can increase later
file_counter = 0
results = []

# =========================
# METRIC FUNCTIONS
# =========================
def count_instructions(ir_file):
    count = 0
    with open(ir_file, 'r') as f:
        for line in f:
            if line.strip().startswith("%"):
                count += 1
    return count

def count_functions(ir_file):
    count = 0
    with open(ir_file, 'r') as f:
        for line in f:
            if line.strip().startswith("define"):
                count += 1
    return count

def count_basic_blocks(dot_file):
    count = 0
    with open(dot_file, 'r') as f:
        for line in f:
            if "label=" in line:
                count += 1
    return count


# =========================
# MAIN PROCESSING LOOP
# =========================
for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        if file.endswith(".c"):

            file_counter += 1
            if file_counter > MAX_FILES:
                break

            c_path = os.path.join(root, file)
            base = str(file_counter)

            ll_file = os.path.join(IR_PATH, base + ".ll")
            bc_file = os.path.join(IR_PATH, base + ".bc")

            print(f"Processing {file_counter}: {file}")

            try:
                # Generate LLVM IR
                subprocess.run(
                    ["clang", "-S", "-emit-llvm", "-O0",
                     "-Xclang", "-disable-O0-optnone",
                     c_path, "-o", ll_file],
                    check=True,
                    stdout=subprocess.DEVNULL,
                    stderr=subprocess.DEVNULL
                )

                # Convert to bitcode
                subprocess.run(
                    ["llvm-as", ll_file, "-o", bc_file],
                    check=True,
                    stdout=subprocess.DEVNULL,
                    stderr=subprocess.DEVNULL
                )

                # Generate CFG (.dot files)
                subprocess.run(
                    ["opt", "-enable-new-pm=0", "-dot-cfg",
                     bc_file, "-disable-output"],
                    check=True,
                    stdout=subprocess.DEVNULL,
                    stderr=subprocess.DEVNULL
                )

                # Collect metrics
                instr_count = count_instructions(ll_file)
                func_count = count_functions(ll_file)

                dot_files = [f for f in os.listdir(".") if f.endswith(".dot")]
                bb_count = 0

                for dot in dot_files:
                    bb_count += count_basic_blocks(dot)
                    os.remove(dot)  # Clean dot files immediately

                results.append({
                    "file": file,
                    "instructions": instr_count,
                    "functions": func_count,
                    "basic_blocks": bb_count
                })
                issues = check_division_by_zero(ll_file)



            except Exception as e:
                print("Skipped:", file)
                continue

    if file_counter > MAX_FILES:
        break

# =========================
# SAVE RESULTS
# =========================
df = pd.DataFrame(results)
df.to_csv("ir_statistics.csv", index=False)

print("\nDONE ✅")
df.head()
````

create IR tool.py

````python
%%writefile ir_tool.py
import os
import sys
import subprocess

# ===============================
# METRIC FUNCTIONS
# ===============================

def count_instructions(ir_file):
    count = 0
    with open(ir_file, 'r') as f:
        for line in f:
            if line.strip().startswith("%"):
                count += 1
    return count

def count_functions(ir_file):
    count = 0
    with open(ir_file, 'r') as f:
        for line in f:
            if line.strip().startswith("define"):
                count += 1
    return count

def count_basic_blocks(dot_file):
    count = 0
    with open(dot_file, 'r') as f:
        for line in f:
            if "label=" in line:
                count += 1
    return count

# ===============================
# IR CLEANER
# ===============================

def clean_ir(input_file, output_file):
    with open(input_file, 'r') as f:
        lines = f.readlines()

    cleaned = []
    for line in lines:
        if line.strip().startswith("!"):
            continue
        if "attributes #" in line:
            continue
        cleaned.append(line)

    with open(output_file, 'w') as f:
        f.writelines(cleaned)

# ===============================
# BASIC VERIFICATION CHECKS
# ===============================

def check_division_by_zero(ir_file):
    issues = []
    with open(ir_file, 'r') as f:
        for line_no, line in enumerate(f, 1):
            if " sdiv " in line or " udiv " in line:
                if ", 0" in line:
                    issues.append(f"Division by zero detected at IR line {line_no}")
    return issues

def check_missing_return(ir_file):
    issues = []
    with open(ir_file, 'r') as f:
        content = f.read()
        if "ret " not in content:
            issues.append("Warning: No return instruction found in IR.")
    return issues

def check_unreachable_blocks(dot_files):
    # Simple heuristic: if CFG has only 1 block but contains branch instruction,
    # something unusual may be happening
    warnings = []
    if len(dot_files) == 0:
        warnings.append("No CFG generated.")
    return warnings

# ===============================
# MAIN FUNCTION
# ===============================

def main():
    if len(sys.argv) != 2:
        print("Usage: python ir_tool.py <input.c>")
        sys.exit(1)

    c_file = sys.argv[1]

    if not os.path.exists(c_file):
        print("Error: File not found.")
        sys.exit(1)

    base = os.path.splitext(c_file)[0]
    ll_file = base + ".ll"
    clean_file = base + "_clean.ll"
    bc_file = base + ".bc"
    cfg_image = base + "_cfg.png"

    print("Generating LLVM IR...")
    subprocess.run(
        ["clang", "-S", "-emit-llvm", "-O0",
         "-Xclang", "-disable-O0-optnone",
         c_file, "-o", ll_file],
        check=True
    )

    print("Cleaning IR...")
    clean_ir(ll_file, clean_file)

    print("Generating bitcode...")
    subprocess.run(["llvm-as", ll_file, "-o", bc_file], check=True)

    print("Generating CFG...")
    subprocess.run(
        ["opt", "-enable-new-pm=0", "-dot-cfg",
         bc_file, "-disable-output"],
        check=True
    )

    dot_files = [f for f in os.listdir(".") if f.endswith(".dot")]
    total_blocks = 0

    for dot in dot_files:
        total_blocks += count_basic_blocks(dot)
        subprocess.run(["dot", "-Tpng", dot, "-o", cfg_image])
        os.remove(dot)

    instr_count = count_instructions(ll_file)
    func_count = count_functions(ll_file)

    # ===============================
    # VERIFICATION SECTION
    # ===============================

    div_zero_issues = check_division_by_zero(ll_file)
    return_issues = check_missing_return(ll_file)
    cfg_warnings = check_unreachable_blocks(dot_files)

    print("\n===== ANALYSIS SUMMARY =====")
    print("Instructions:", instr_count)
    print("Functions:", func_count)
    print("Basic Blocks:", total_blocks)
    print("Clean IR saved as:", clean_file)
    print("CFG image saved as:", cfg_image)

    print("\n===== BASIC VERIFICATION RESULTS =====")

    if not div_zero_issues and not return_issues and not cfg_warnings:
        print("No basic verification issues detected.")
    else:
        for issue in div_zero_issues:
            print(issue)
        for issue in return_issues:
            print(issue)
        for issue in cfg_warnings:
            print(issue)

if __name__ == "__main__":
    main()
````
